```python
import re
import matplotlib.pyplot as plt
import pandas as pd
```

```python
log_path = "/mnt/transfer/TEbenchmarking/logs/stellarscope/"
# 1. Read the log file
# Replace 'stellarscope.log' with the actual path to your log file
dataset = "simulated_mm_RA"
sample = "young"
runid = "_260519_seed213"

with open(log_path + dataset+ "_" + sample + "_resume_pseudobulk_manualrun" + runid + ".txt", 'r') as f:
    log_text = f.read()

# 2. Extract Iteration and Diff values using Regular Expressions
pattern = r"Iteration (\d+), lnl= ([\d.e+-]+), diff=([\d.e+-]+)"
matches = re.findall(pattern, log_text)

# 3. Convert to a Pandas DataFrame for easy handling
df = pd.DataFrame(matches, columns=["Iteration", "Lnl", "Diff"])
df["Iteration"] = df["Iteration"].astype(int)
df["Lnl"] = df["Lnl"].astype(float)
df["Diff"] = df["Diff"].astype(float)

# 4. Create the plot
plt.figure(figsize=(10, 6))
plt.plot(df["Iteration"], df["Diff"], marker='o', color='#1f77b4', linestyle='-', linewidth=2, label="Parameter Change ($\Delta\pi$)")

# Style the plot
plt.title("Stellarscope EM Convergence Curve (Parameter Stability) - Scenario 1 young", fontsize=14, fontweight='bold')
plt.xlabel("Iteration", fontsize=12)
plt.ylabel("Absolute Difference (diff)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.yscale('log') # EM differences drop exponentially; a log scale makes it easy to read
plt.legend()

# Show or save the plot
plt.tight_layout()
#plt.savefig("em_convergence_curve.png", dpi=300)
plt.show()


# 4. Create the plot
plt.figure(figsize=(10, 6))
plt.plot(df["Iteration"], df["Lnl"], marker='o', color='#1f77b4', linestyle='-', linewidth=2)

# Style the plot
plt.title("Stellarscope EM - lnl in Scenario 1 young", fontsize=14, fontweight='bold')
plt.xlabel("Iteration", fontsize=12)
plt.ylabel("lnl", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.yscale('log') # EM differences drop exponentially; a log scale makes it easy to read
plt.legend()

# Show or save the plot
plt.tight_layout()
#plt.savefig("em_convergence_curve.png", dpi=300)
plt.show()
```
